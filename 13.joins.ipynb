```python
import pandas as pd 
import numpy as np
```

```python
bat = pd.read_csv("batsman_runs_ipl - batsman_runs_ipl.csv")
bat
```

```python
top_batsmen = bat.sort_values("batsman_run", ascending=False).head(5).reset_index(drop=True)
print(top_batsmen)

print()

player_team = pd.DataFrame({
    'batter' : ['V Kohli', 'S Dhawan', 'DA Warner', 'RG Sharma', 'MS Dhoni'],
    'Team' : ['RCB', 'DC', 'SRH', 'MI', 'CSK']

})

print(player_team)
```

```python
pd.merge(top_batsmen, player_team, on="batter", how="inner")
```

```python
pd.merge(top_batsmen, player_team, on="batter", how="left")
```

```python
pd.merge(top_batsmen, player_team, on="batter", how="outer")
```

```python
player_team_v2 = player_team.rename(columns={'batter':'Player Name'})
```

```python
player_team_v2
```

```python
pd.merge(top_batsmen, player_team_v2, left_on='batter', right_on='Player Name', how='inner')
```
