```python
df["Revenue"]
```

```python
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

Interpret Revenue Skew

```python
if revenue_skew> 0:
    print("Positive Skew")
elif revenue_skew< 0 :
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

```python
unit_skew = df["Units_Sold"].skew()
print(unit_skew)
```

# Matplot library

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df["Revenue"], bins=20)
plt.title("Revenue Dirstribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["Units_Sold"], bins=20)
plt.title("Unit Sold Distribution")
plt.xlabel("Unit Sold")
plt.ylabel("Frequency")
plt.show()
```
