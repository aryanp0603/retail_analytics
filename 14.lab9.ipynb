```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
n = 20
p = 0.30
```

```python
print("Numbers of customers", n)
print("Purchase Probability", p)
```

```python
prob_5 = binom.pmf(5, n, p)
print(prob_5)
```

```python
prob_6 = binom.pmf(6, n, p)
print(prob_6)
```

```python
prob_10 = binom.pmf(10, n, p)
print(prob_10)
```

```python
x = np.arange(0, n+1)
```

```python
y = binom.pmf(x, n, p)
```

```python
probability_table = pd.DataFrame({
    "Purchases": x,
    "Probability": y
})

probability_table.head(10)
```

```python
plt.figure(figsize=(10,5))
plt.bar(x, y)
plt.title("Binomial Distribution")
plt.xlabel("Number of Purchases")
plt.ylabel("Probability")
plt.show()
```

```python
most_likely = x[np.argmax(y)]
print("Most Likely Purchases:", most_likely)
```

```python
expected_purchases = n * p
print(expected_purchases)
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Number of Customers",
        "Purchase Probability",
        "Expected Purchases",
        "Most Likely Purchases"
    ],
    "Value": [
        n,
        p,
        expected_purchases,
        most_likely
    ]
})

summary
```
