```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Retail_Dataset_Cleaned.csv")
df
```

```python
df.columns
```

```python
monthly_revenue = df.groupby("Month")["Revenue"].sum()\

plt.figure(figsize=(10, 5))
```

```python
plt.plot(monthly_revenue.index,
         monthly_revenue.values,
         marker='o',
         linewidth=2)

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
category_revenue = df.groupby("Category")["Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))

plt.bar(category_revenue.index,
         category_revenue.values
         )

plt.title('Category Revenue Trend')
plt.xlabel('Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.hist(df['Revenue'],
         bins=20)

plt.title('Distribution of Revenue')
plt.xlabel('Revenue')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Units_Sold'],
            df['Revenue'])

plt.title('Units Sold vs Revenue')
plt.xlabel('Units Sold')
plt.ylabel('Revenue')

plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.boxplot(df['Profit_Margin'])
plt.title('Profit Margin Distribution')
plt.ylabel('Profit Margin')
plt.show()
```

```python
category_revenue = df.groupby('Category')['Revenue'].sum()

# Create Pie Chart
plt.figure(figsize=(8, 8))

plt.pie(category_revenue,
        labels=category_revenue.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title('Revenue Distribution by Category')

plt.show()
```

```python
bt = pd.read_csv("batsman_season_record - batsman_season_record.csv")
bt
```

```python
plt.bar(bt['batsman'], bt['2015'], color= 'magenta')
plt.bar(bt['batsman'], bt['2016'], color= 'blue')
plt.bar(bt['batsman'], bt['2017'], color= 'black')
plt.legend(['2015', '2016', '2017'])
```

```python
children = [10, 20, 40, 10, 30]
colors = ['red red red red red', 'blue blue blue blue blue', 'green green green green green', 'yellow yellow yellow yellow yellow', 'pink pink pink pink pink']
plt.bar(colors, children, color='black')
plt.xticks(rotation='vertical')
```

```python
plt.bar(bt['batsman'], bt['2017'], label = '2017')
plt.bar(bt['batsman'], bt['2016'], bottom= bt['2017'], label='2016')
plt.bar(bt['batsman'], bt['2015'], bottom=(bt['2016'] + bt['2017']), label= '2015')

plt.legend()
plt.show()
```
