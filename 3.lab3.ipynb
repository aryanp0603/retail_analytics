```python
high_price = rd[rd['Unit_Price']>1000]
high_price[['Store_Name', 'Brand_Name', 'Category']].head()
```

```python
low_price = rd[rd['Unit_Price']<500]
low_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
medium_price = rd[
    (rd['Unit_Price'] >= 500) &
    (rd['Unit_Price'] <= 1500)
]

medium_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
sorted_price = rd.sort_values(
    by = 'Unit_Price',
    ascending=False
)
sorted_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
top_price = rd.nlargest(10, 'Unit_Price')
print(top_price[['Unit_Price']])
```
