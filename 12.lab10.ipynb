```python
city_rev = pd.pivot_table(
    retail,
    values= "Revenue",
    index= "City",
    aggfunc="sum"
)
```

```python
segment_rev = pd.pivot_table(
    retail,
    values="Revenue",
    index="Customer_Segment",
    aggfunc="sum"
)

segment_rev
```

```python
category_summary = pd.pivot_table(
    retail,
    values= "Revenue",
    index= "Category",
    aggfunc=["sum", "count", "mean", "max", "min"]
)

category_summary
```

```python
city_performance = pd.pivot_table(
    retail,
    values=["Revenue", "Profit_Margin"],
    index= "City",
    aggfunc={
        "Revenue" : "sum",
        "Profit_Margin" : "mean"
    })
```

```python
city_performance
```
