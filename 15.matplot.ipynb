```python
batsman = pd.read_csv("sharma-kohli - sharma-kohli.csv")
batsman
```

```python
plt.plot(batsman["index"], batsman["V Kohli"])
plt.plot(batsman["index"], batsman["RG Sharma"])
```

```python
plt.plot(batsman["index"], batsman["V Kohli"])
plt.plot(batsman["index"], batsman["RG Sharma"])

plt.title("Rohit Sharma VS Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Run Scored")
```

```python
plt.plot(batsman["index"], batsman["V Kohli"], color='black')
plt.plot(batsman["index"], batsman["RG Sharma"], color= 'red')

plt.title("Rohit Sharma VS Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Run Scored")
```

```python
plt.plot(batsman["index"], batsman["V Kohli"], color='black', linestyle='solid', linewidth=6)
plt.plot(batsman["index"], batsman["RG Sharma"], color= 'red', linestyle='dashdot', linewidth=5)

plt.title("Rohit Sharma VS Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Run Scored")
```

```python
plt.plot(batsman["index"], batsman["V Kohli"], color='black', linestyle='solid', linewidth=3, marker='o', markersize=8)
plt.plot(batsman["index"], batsman["RG Sharma"], color= 'red', linestyle='dashdot', linewidth=3, marker='*', markersize= 8)

plt.title("Rohit Sharma VS Virat Kohli Career Comparison")
plt.xlabel("Season")
plt.ylabel("Run Scored")
```
